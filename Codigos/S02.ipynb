### Carga de modulos o clases

```python
# Estructura de llamado o carga de modulo 

# import nombre_modulo - Importar el modulo completo sin abreviatura 

import numpy 
```

```python
numpy.mean([10,20,30,40])
```

```python
# import nombre_modulo as abreviatura - Importar el modulo completo con abreviatura 

import numpy as np
```

```python
np.mean([10, 20, 30, 40])
```

```python
# from nombre_modulo import clase/submodulo/herencia

from numpy import mean
```

```python
mean([10, 20, 30, 40])
```

En ocasiones, hay modulos que contienen otros *submodulos* que a su vez dentro de cada *submodulo* hay diferentes *clases* (funciones), por lo tanto no se pueden llamar desde el modulo general, ejemplo:

```python
import statsmodels.api as sm
```

```python
sm.jarque_bera([10, 20, 30, 40])
```

Dentro de statsmodels (que esta abreviado como sm), sí existe la clase *jarque_bera* pero para ser usada hay llamar primero al submodulo de *stats* y de *statstools*, entonces quedaría de la siguiente manera:

```python
# modulo.submodulo1.submodulo2.clase 
sm.stats.stattools.jarque_bera([10, 20, 30, 40])
```

```python
# Llamando directamente a la clase 
from statsmodels.stats.stattools import jarque_bera
```

```python
jarque_bera([10, 20, 30, 40])
```

### Array 

```python
import numpy as np # Es necesario tener numpy cargado 

Edades = np.array([24, 34, 40, 28, 27, 19, 20, 37])
Edades
```

El objeto array permite realizar operaciones internas (desde el objeto) tales como: Suma, media, máximo, minimo, etc.

```python
Edades.mean()
```

```python
Edades.sum()
```

A diferencia de la lista que no define esos atributos dentro del obejeto.

```python
Edades2 = [24, 34, 40, 28, 27, 19, 20, 37]
Edades2.mean()
```

```python
# Elementos o atributos del array 

dir(Edades)
```

#### Matrices con array 

```python
# Matriz de elementos 

Matriz = np.array(
    [
        [11, 21, 31], #Fila 1
        [21, 22, 23], #Fila 2
        [31, 32, 33], #Fila 3
    ]
)

Matriz
```

```python
Matriz.T
```

```python
Matriz.T * Matriz
```

```python
# Inversa 
A = np.array(
    [
        [2, 5, 9], #Fila 1
        [11, 3, 7], #Fila 2
        [8, 4, 12], #Fila 3
    ]
)


np.linalg.inv(A)
```

```python
# Determinante 

np.linalg.det(A)
```
